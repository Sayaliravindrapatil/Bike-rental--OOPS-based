```python
from nltk.text import TokenSearcher
from nltk.sem.logic import Tokens
from nltk.corpus import wordnet

def get_wordnet_pos(i):
  # for i in my_dict.values():
    if i.startswith('J'):
        return wordnet.ADJ
    elif i.startswith('V'):
        return wordnet.VERB
    elif i.startswith('N'):
        return wordnet.NOUN
    elif i.startswith('R'):
        return wordnet.ADV
    else:
        return ''

```
